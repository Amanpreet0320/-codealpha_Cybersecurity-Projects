## 4.10 Secure Password Hashing Service

A dedicated password service is implemented using BCrypt to securely hash user passwords and verify submitted passwords against their stored hashes. This prevents plaintext password storage and provides secure password verification.

````python
%%writefile /content/CodeAlpha_Secure_Coding/src/main/java/com/example/demo/service/PasswordService.java
package com.example.demo.service;

import org.springframework.security.crypto.bcrypt.BCryptPasswordEncoder;
import org.springframework.stereotype.Service;

@Service
public class PasswordService {

    private final BCryptPasswordEncoder encoder = new BCryptPasswordEncoder();

    public String hashPassword(String password) {
        return encoder.encode(password);
    }

    public boolean verifyPassword(String password, String hashedPassword) {
        return encoder.matches(password, hashedPassword);
    }
}
````

## 4.13 Authentication Controller Implementation

An authentication controller is implemented to provide registration and login endpoints. The controller validates required credentials, prevents duplicate usernames, securely hashes passwords using the password service, and verifies passwords during login without storing plaintext passwords.

````python
%%writefile /content/CodeAlpha_Secure_Coding/src/main/java/com/example/demo/controller/AuthController.java
package com.example.demo.controller;

import com.example.demo.model.User;
import com.example.demo.service.PasswordService;
import org.springframework.web.bind.annotation.*;

import java.util.HashMap;
import java.util.Map;

@RestController
@RequestMapping("/auth")
public class AuthController {

    private final PasswordService passwordService;
    private final Map<String, String> users = new HashMap<>();

    public AuthController(PasswordService passwordService) {
        this.passwordService = passwordService;
    }

    @PostMapping("/register")
    public String register(@RequestBody User user) {

        if (user.getUsername() == null || user.getUsername().isBlank()
                || user.getPassword() == null || user.getPassword().isBlank()) {
            return "Username and password are required";
        }

        if (users.containsKey(user.getUsername())) {
            return "Username already exists";
        }

        String hashedPassword = passwordService.hashPassword(user.getPassword());
        users.put(user.getUsername(), hashedPassword);

        return "Registration successful";
    }

    @PostMapping("/login")
    public String login(@RequestBody User user) {

        String hashedPassword = users.get(user.getUsername());

        if (hashedPassword != null
                && passwordService.verifyPassword(user.getPassword(), hashedPassword)) {
            return "Login successful";
        }

        return "Invalid username or password";
    }
}
````

## 4.15 Password Security Unit Test

A unit test is added to verify the password security service. The test confirms that passwords are not stored in plaintext, valid passwords can be verified against their BCrypt hashes, and incorrect passwords are rejected.

````python
%%writefile /content/CodeAlpha_Secure_Coding/src/test/java/com/example/demo/PasswordServiceTest.java
package com.example.demo;

import com.example.demo.service.PasswordService;
import org.junit.jupiter.api.Test;

import static org.junit.jupiter.api.Assertions.*;

class PasswordServiceTest {

    @Test
    void passwordShouldBeHashedAndVerified() {
        PasswordService service = new PasswordService();

        String password = "SecurePassword123!";
        String hashedPassword = service.hashPassword(password);

        assertNotEquals(password, hashedPassword);
        assertTrue(service.verifyPassword(password, hashedPassword));
        assertFalse(service.verifyPassword("WrongPassword", hashedPassword));
    }
}
````
