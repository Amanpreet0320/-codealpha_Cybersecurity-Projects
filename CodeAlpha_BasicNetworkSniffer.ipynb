## How the Network Sniffer Works

The network sniffer captures packets flowing through the available network interface using the Scapy `sniff()` function.

For each captured packet, the program checks whether IP information is available and extracts:

- Source IP address
- Destination IP address
- Network protocol
- Source port
- Destination port
- Packet length
- Raw payload, when available

The captured packets are then analyzed to calculate protocol statistics and are exported to a CSV file for further analysis.

## Step 2: Capture Network Packets

Scapy's `sniff()` function is used to capture network packets. In the initial capture, 20 packets are collected and the total number of captured packets is displayed.

```python
packets = sniff(count=20)

print("Packets captured:", len(packets))
```

## Step 3: Packet Summary

The captured packets are summarized using Scapy's `summary()` function. This provides a quick overview of the packet structure, including the network layers, source and destination addresses, ports, and protocol information.

```python
packets.summary()
```

## Step 4: Analyze IP Addresses and Protocols

Each captured packet is examined to identify its source IP address, destination IP address, and network protocol.

The program checks whether the packet contains an IP layer. It then identifies whether the packet uses TCP, UDP, or another protocol and displays the relevant network information.

```python
for i, packet in enumerate(packets, start=1):
    print(f"\nPacket {i}")

    if IP in packet:
        print("Source IP      :", packet[IP].src)
        print("Destination IP :", packet[IP].dst)

        if TCP in packet:
            print("Protocol       : TCP")
        elif UDP in packet:
            print("Protocol       : UDP")
        else:
            print("Protocol       : Other")
    else:
        print("No IP information available")
```

```python
for i, packet in enumerate(packets, start=1):
    print(f"\n========== Packet {i} ==========")

    if IP in packet:
        print("Source IP       :", packet[IP].src)
        print("Destination IP  :", packet[IP].dst)

        if TCP in packet:
            print("Protocol        : TCP")
            print("Source Port     :", packet[TCP].sport)
            print("Destination Port:", packet[TCP].dport)

        elif UDP in packet:
            print("Protocol        : UDP")
            print("Source Port     :", packet[UDP].sport)
            print("Destination Port:", packet[UDP].dport)

        else:
            print("Protocol        : Other")

        print("Packet Length   :", len(packet), "bytes")

    else:
        print("No IP information available")
```

## Payload Analysis

This section examines the raw payload data present in the captured packets and reports the payload length and a short preview for packets containing readable raw data.

```python
print("=" * 60)
print("             PAYLOAD ANALYSIS")
print("=" * 60)

payload_count = 0

for i, packet in enumerate(packets, start=1):

    if Raw in packet:
        payload = packet[Raw].load
        payload_count += 1

        print(f"\nPacket {i}")
        print("Payload Length:", len(payload), "bytes")
        print("Payload Preview:", payload[:50])

print("\nTotal packets containing a Raw payload:", payload_count)
```

## Step 5: Protocol Statistics

The captured packets are analyzed to determine the number of packets belonging to different network protocols. The program counts TCP, UDP, and other packets to provide a summary of the captured network traffic.

```python
tcp_count = sum(1 for packet in packets if TCP in packet)
udp_count = sum(1 for packet in packets if UDP in packet)
other_count = len(packets) - tcp_count - udp_count

print("Protocol Statistics")
print("-" * 30)
print(f"TCP   : {tcp_count}")
print(f"UDP   : {udp_count}")
print(f"Other : {other_count}")
print(f"Total : {len(packets)}")
```

## Step 6: Visualize Protocol Statistics

The protocol statistics are visualized using a bar chart with Matplotlib. The chart compares the number of TCP, UDP, and other packets captured during the network sniffing process. This provides a simple visual representation of the protocol distribution in the captured traffic.

```python
import matplotlib.pyplot as plt

protocols = ["TCP", "UDP", "Other"]
counts = [tcp_count, udp_count, other_count]

plt.figure(figsize=(7, 5))
bars = plt.bar(protocols, counts)

plt.title("Network Packet Protocol Statistics")
plt.xlabel("Protocol")
plt.ylabel("Number of Packets")

# Display values above each bar
for bar, count in zip(bars, counts):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        count + 0.3,
        str(count),
        ha="center"
    )

plt.tight_layout()
plt.show()
```

## Step 7: Export Packet Data to CSV

The analyzed packet information is stored in a CSV file for further analysis and record keeping. The CSV file contains the packet number, source and destination IP addresses, protocol, source and destination ports, and packet length. This makes the captured network data easier to review and analyze using data analysis tools.

```python
import csv

with open("network_packets.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow([
        "Packet No",
        "Source IP",
        "Destination IP",
        "Protocol",
        "Source Port",
        "Destination Port",
        "Packet Length"
    ])

    for i, packet in enumerate(packets, start=1):

        if IP in packet:

            source_ip = packet[IP].src
            destination_ip = packet[IP].dst
            packet_length = len(packet)

            if TCP in packet:
                protocol = "TCP"
                source_port = packet[TCP].sport
                destination_port = packet[TCP].dport

            elif UDP in packet:
                protocol = "UDP"
                source_port = packet[UDP].sport
                destination_port = packet[UDP].dport

            else:
                protocol = "Other"
                source_port = "-"
                destination_port = "-"

            writer.writerow([
                i,
                source_ip,
                destination_ip,
                protocol,
                source_port,
                destination_port,
                packet_length
            ])

print("CSV file created successfully!")
```
